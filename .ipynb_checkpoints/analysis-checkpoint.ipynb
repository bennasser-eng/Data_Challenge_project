```python
import pandas as pd
import datetime
import holidays

# public holiday
FR_HOLIDAYS = holidays.France()

# the number of business hours is 8 
def get_business_hours(start, end, h_open=8, h_close=17):
    """
    the business time between start & end.
    """
    if pd.isna(start) or pd.isna(end) or end < start:
        return 0.0

    # Make start & end in [8h, 17h]
    # (e.g: 06h30 -> 8.0, 19h00 -> 17.0)
    t_start = max(h_open, min(h_close, start.hour + start.minute / 60.0))
    t_end = max(h_open, min(h_close, end.hour + end.minute / 60.0))

    # 1 case : The tiket is resolve same day:
    if start.date() == end.date():
        if start.weekday() >= 5 or start.date() in FR_HOLIDAYS:
            return 0.0
        return max(0.0, t_end - t_start)

    # 2 case : Tiket is resolve under many days
    total = 0.0

    # A. Hours of the first day (from stat to 17h)
    if start.weekday() < 5 and start.date() not in FR_HOLIDAYS:
        total += (h_close - t_start)

    # B. Fours of the last day (from 08h to end)
    if end.weekday() < 5 and end.date() not in FR_HOLIDAYS:
        total += (t_end - h_open)

    # C. All days betwenn first day and last day in weekdays and without public holidays
    day_step = datetime.timedelta(days=1)
    current_day = start.date() + day_step
    daily_hours = h_close - h_open      # 9 hours

    while current_day < end.date():
        # We add 9h for weekdays (no public days)
        if current_day.weekday() < 5 and current_day not in FR_HOLIDAYS:
            total += daily_hours
        current_day += day_step

    return total
```

```python
# Conversion to datetime
df['SUBMIT_DATE_UT'] = pd.to_datetime(df['SUBMIT_DATE_UT'])
df['END_DATE_UT'] = pd.to_datetime(df['END_DATE_UT'])

# Copute of the taget_value for régression
df['target_regression'] = df.apply(
    lambda row: get_business_hours(row['SUBMIT_DATE_UT'], row['END_DATE_UT']), 
    axis=1
)

df.head()
```

```python
df['END_DATE_UT'] = pd.to_datetime(df['END_DATE_UT'])
df['MAX_RESOLUTION_DATE_UT'] = pd.to_datetime(df['MAX_RESOLUTION_DATE_UT'])

# Target_classification (1 = SLA Violé, 0 = SLA Respecté)
df['target_classification'] = (df['END_DATE_UT'] > df['MAX_RESOLUTION_DATE_UT']).astype(int)

df.head()
```

```python
print("Distribution brute :")
print(df['target_classification'].value_counts(dropna=False))

print("\nProportion (%) :")
print(df['target_classification'].value_counts(normalize=True) * 100)
```
